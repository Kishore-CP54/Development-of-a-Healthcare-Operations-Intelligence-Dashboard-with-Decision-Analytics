## Question 1: SELECT and WHERE

**Problem Statement:** Retrieve the names and ages of all employees who work in the 'Sales' department.

**Sample Table: Employees**

| EmployeeID | Name    | Age | Department | Salary |
|------------|---------|-----|------------|--------|
| 1          | Alice   | 30  | Sales      | 60000  |
| 2          | Bob     | 24  | Marketing  | 50000  |
| 3          | Charlie | 35  | Sales      | 75000  |
| 4          | David   | 28  | IT         | 62000  |

**Expected Output:**

| Name    | Age |
|---------|-----|
| Alice   | 30  |
| Charlie | 35  |

```python
import sqlite3
import pandas as pd

# Connect to an in-memory SQLite database
conn = sqlite3.connect(':memory:')
cursor = conn.cursor()

# Create the Employees table
cursor.execute('''
CREATE TABLE Employees (
    EmployeeID INTEGER PRIMARY KEY,
    Name TEXT,
    Age INTEGER,
    Department TEXT,
    Salary INTEGER
);
''')

# Insert sample data
cursor.execute("INSERT INTO Employees VALUES (1, 'Alice', 30, 'Sales', 60000);")
cursor.execute("INSERT INTO Employees VALUES (2, 'Bob', 24, 'Marketing', 50000);")
cursor.execute("INSERT INTO Employees VALUES (3, 'Charlie', 35, 'Sales', 75000);")
cursor.execute("INSERT INTO Employees VALUES (4, 'David', 28, 'IT', 62000);")
conn.commit()

# SQL Query
sql_query = "SELECT Name, Age FROM Employees WHERE Department = 'Sales';"

# Execute query and fetch results
cursor.execute(sql_query)
results = cursor.fetchall()

# Display results using pandas for better formatting
df = pd.DataFrame(results, columns=['Name', 'Age'])
display(df)

# Explanation
print("\nExplanation: The `SELECT` statement specifies the columns `Name` and `Age` to retrieve. The `FROM` clause indicates the `Employees` table. The `WHERE` clause filters the rows, ensuring only employees from the 'Sales' department are included in the result.")

# Close the connection
conn.close()
```

## Question 2: ORDER BY

**Problem Statement:** List all products, ordered by their price from lowest to highest.

**Sample Table: Products**

| ProductID | ProductName | Price | Category |
|-----------|-------------|-------|----------|
| 101       | Laptop      | 1200  | Electronics |
| 102       | Mouse       | 25    | Electronics |
| 103       | Keyboard    | 75    | Electronics |
| 104       | Monitor     | 300   | Electronics |

**Expected Output:**

| ProductID | ProductName | Price | Category |
|-----------|-------------|-------|----------|
| 102       | Mouse       | 25    | Electronics |
| 103       | Keyboard    | 75    | Electronics |
| 104       | Monitor     | 300   | Electronics |
| 101       | Laptop      | 1200  | Electronics |

```python
import sqlite3
import pandas as pd

# Connect to an in-memory SQLite database
conn = sqlite3.connect(':memory:')
cursor = conn.cursor()

# Create the Products table
cursor.execute('''
CREATE TABLE Products (
    ProductID INTEGER PRIMARY KEY,
    ProductName TEXT,
    Price INTEGER,
    Category TEXT
);
''')

# Insert sample data
cursor.execute("INSERT INTO Products VALUES (101, 'Laptop', 1200, 'Electronics');")
cursor.execute("INSERT INTO Products VALUES (102, 'Mouse', 25, 'Electronics');")
cursor.execute("INSERT INTO Products VALUES (103, 'Keyboard', 75, 'Electronics');")
cursor.execute("INSERT INTO Products VALUES (104, 'Monitor', 300, 'Electronics');")
conn.commit()

# SQL Query
sql_query = "SELECT ProductID, ProductName, Price, Category FROM Products ORDER BY Price ASC;"

# Execute query and fetch results
cursor.execute(sql_query)
results = cursor.fetchall()

# Display results using pandas
df = pd.DataFrame(results, columns=['ProductID', 'ProductName', 'Price', 'Category'])
display(df)

# Explanation
print("\nExplanation: The `ORDER BY Price ASC` clause sorts the result set by the `Price` column in ascending order (lowest to highest). `ASC` is optional as it's the default order.")

# Close the connection
conn.close()
```

## Question 3: GROUP BY and COUNT

**Problem Statement:** Count the number of employees in each department.

**Sample Table: Employees**

| EmployeeID | Name    | Department |
|------------|---------|------------|
| 1          | Alice   | Sales      |
| 2          | Bob     | Marketing  |
| 3          | Charlie | Sales      |
| 4          | David   | IT         |
| 5          | Eve     | Marketing  |

**Expected Output:**

| Department | EmployeeCount |
|------------|---------------|
| IT         | 1             |
| Marketing  | 2             |
| Sales      | 2             |

```python
import sqlite3
import pandas as pd

# Connect to an in-memory SQLite database
conn = sqlite3.connect(':memory:')
cursor = conn.cursor()

# Create the Employees table
cursor.execute('''
CREATE TABLE Employees (
    EmployeeID INTEGER PRIMARY KEY,
    Name TEXT,
    Department TEXT
);
''')

# Insert sample data
cursor.execute("INSERT INTO Employees VALUES (1, 'Alice', 'Sales');")
cursor.execute("INSERT INTO Employees VALUES (2, 'Bob', 'Marketing');")
cursor.execute("INSERT INTO Employees VALUES (3, 'Charlie', 'Sales');")
cursor.execute("INSERT INTO Employees VALUES (4, 'David', 'IT');")
cursor.execute("INSERT INTO Employees VALUES (5, 'Eve', 'Marketing');")
conn.commit()

# SQL Query
sql_query = "SELECT Department, COUNT(EmployeeID) AS EmployeeCount FROM Employees GROUP BY Department;"

# Execute query and fetch results
cursor.execute(sql_query)
results = cursor.fetchall()

# Display results using pandas
df = pd.DataFrame(results, columns=['Department', 'EmployeeCount'])
display(df)

# Explanation
print("\nExplanation: The `GROUP BY Department` clause groups rows that have the same department. For each group, `COUNT(EmployeeID)` calculates the total number of employees. `AS EmployeeCount` renames the count column for clarity.")

# Close the connection
conn.close()
```

## Question 4: Aggregate Functions (MAX and MIN)

**Problem Statement:** Find the highest and lowest salaries among all employees.

**Sample Table: Employees**

| EmployeeID | Name    | Salary |
|------------|---------|--------|
| 1          | Alice   | 60000  |
| 2          | Bob     | 50000  |
| 3          | Charlie | 75000  |
| 4          | David   | 62000  |

**Expected Output:**

| MaxSalary | MinSalary |
|-----------|-----------|
| 75000     | 50000     |

```python
import sqlite3
import pandas as pd

# Connect to an in-memory SQLite database
conn = sqlite3.connect(':memory:')
cursor = conn.cursor()

# Create the Employees table
cursor.execute('''
CREATE TABLE Employees (
    EmployeeID INTEGER PRIMARY KEY,
    Name TEXT,
    Salary INTEGER
);
''')

# Insert sample data
cursor.execute("INSERT INTO Employees VALUES (1, 'Alice', 60000);")
cursor.execute("INSERT INTO Employees VALUES (2, 'Bob', 50000);")
cursor.execute("INSERT INTO Employees VALUES (3, 'Charlie', 75000);")
cursor.execute("INSERT INTO Employees VALUES (4, 'David', 62000);")
conn.commit()

# SQL Query
sql_query = "SELECT MAX(Salary) AS MaxSalary, MIN(Salary) AS MinSalary FROM Employees;"

# Execute query and fetch results
cursor.execute(sql_query)
results = cursor.fetchall()

# Display results using pandas
df = pd.DataFrame(results, columns=['MaxSalary', 'MinSalary'])
display(df)

# Explanation
print("\nExplanation: `MAX(Salary)` returns the highest value in the `Salary` column, and `MIN(Salary)` returns the lowest value. These are aggregate functions that operate on all rows of the table when no `GROUP BY` clause is present.")

# Close the connection
conn.close()
```

| EmployeeID | Name    | Age |
|------------|---------|-----|
| 1          | Alice   | 30  |
| 2          | Bob     | 24  |
| 3          | Charlie | 35  |

**Expected Output:**

## Question 7: HAVING

**Problem Statement:** List departments that have more than 1 employee.

**Sample Table: Employees**

| EmployeeID | Name    | Department |
|------------|---------|------------|
| 1          | Alice   | Sales      |
| 2          | Bob     | Marketing  |
| 3          | Charlie | Sales      |
| 4          | David   | IT         |
| 5          | Eve     | Marketing  |

**Expected Output:**

| Department |
|------------|
| Marketing  |
| Sales      |

```python
import sqlite3
import pandas as pd

# Connect to an in-memory SQLite database
conn = sqlite3.connect(':memory:')
cursor = conn.cursor()

# Create the Employees table
cursor.execute('''
CREATE TABLE Employees (
    EmployeeID INTEGER PRIMARY KEY,
    Name TEXT,
    Department TEXT
);
''')

# Insert sample data
cursor.execute("INSERT INTO Employees VALUES (1, 'Alice', 'Sales');")
cursor.execute("INSERT INTO Employees VALUES (2, 'Bob', 'Marketing');")
cursor.execute("INSERT INTO Employees VALUES (3, 'Charlie', 'Sales');")
cursor.execute("INSERT INTO Employees VALUES (4, 'David', 'IT');")
cursor.execute("INSERT INTO Employees VALUES (5, 'Eve', 'Marketing');")
conn.commit()

# SQL Query
sql_query = "SELECT Department FROM Employees GROUP BY Department HAVING COUNT(EmployeeID) > 1;"

# Execute query and fetch results
cursor.execute(sql_query)
results = cursor.fetchall()

# Display results using pandas
df = pd.DataFrame(results, columns=['Department'])
display(df)

# Explanation
print("\nExplanation: The `GROUP BY Department` clause aggregates rows by department. The `HAVING COUNT(EmployeeID) > 1` clause then filters these grouped results, keeping only those departments where the employee count is greater than 1. `HAVING` is used to filter groups, while `WHERE` filters individual rows.")

# Close the connection
conn.close()
```

## Question 8: JOIN (INNER JOIN)

**Problem Statement:** Retrieve the order ID, customer name, and order date for all orders.

**Sample Table 1: Customers**

| CustomerID | CustomerName |
|------------|--------------|
| 1          | Alice        |
| 2          | Bob          |
| 3          | Charlie      |

**Sample Table 2: Orders**

| OrderID | CustomerID | OrderDate  |
|---------|------------|------------|
| 101     | 1          | 2023-01-15 |
| 102     | 2          | 2023-01-17 |
| 103     | 1          | 2023-01-20 |
| 104     | 3          | 2023-01-22 |

**Expected Output:**

| OrderID | CustomerName | OrderDate  |
|---------|--------------|------------|
| 101     | Alice        | 2023-01-15 |
| 102     | Bob          | 2023-01-17 |
| 103     | Alice        | 2023-01-20 |
| 104     | Charlie      | 2023-01-22 |

```python
import sqlite3
import pandas as pd

# Connect to an in-memory SQLite database
conn = sqlite3.connect(':memory:')
cursor = conn.cursor()

# Create Customers table
cursor.execute('''
CREATE TABLE Customers (
    CustomerID INTEGER PRIMARY KEY,
    CustomerName TEXT
);
''')

# Insert sample Customers data
cursor.execute("INSERT INTO Customers VALUES (1, 'Alice');")
cursor.execute("INSERT INTO Customers VALUES (2, 'Bob');")
cursor.execute("INSERT INTO Customers VALUES (3, 'Charlie');")

# Create Orders table
cursor.execute('''
CREATE TABLE Orders (
    OrderID INTEGER PRIMARY KEY,
    CustomerID INTEGER,
    OrderDate TEXT
);
''')

# Insert sample Orders data
cursor.execute("INSERT INTO Orders VALUES (101, 1, '2023-01-15');")
cursor.execute("INSERT INTO Orders VALUES (102, 2, '2023-01-17');")
cursor.execute("INSERT INTO Orders VALUES (103, 1, '2023-01-20');")
cursor.execute("INSERT INTO Orders VALUES (104, 3, '2023-01-22');")
conn.commit()

# SQL Query
sql_query = """
SELECT
    o.OrderID,
    c.CustomerName,
    o.OrderDate
FROM
    Orders o
INNER JOIN
    Customers c ON o.CustomerID = c.CustomerID;
"""

# Execute query and fetch results
cursor.execute(sql_query)
results = cursor.fetchall()

# Display results using pandas
df = pd.DataFrame(results, columns=['OrderID', 'CustomerName', 'OrderDate'])
display(df)

# Explanation
print("\nExplanation: The `INNER JOIN` clause combines rows from the `Orders` (aliased as `o`) and `Customers` (aliased as `c`) tables where the `CustomerID` matches in both tables. This links each order to its corresponding customer name.")

# Close the connection
conn.close()
```

## Question 9: Finding Duplicate Records

**Problem Statement:** Find all employees whose names appear more than once in the `Employees` table.

**Sample Table: Employees**

| EmployeeID | Name    | Department |
|------------|---------|------------|
| 1          | Alice   | Sales      |
| 2          | Bob     | Marketing  |
| 3          | Alice   | IT         |
| 4          | David   | Sales      |
| 5          | Bob     | Finance    |

**Expected Output:**

| Name |
|------|
| Alice|
| Bob  |

```python
import sqlite3
import pandas as pd

# Connect to an in-memory SQLite database
conn = sqlite3.connect(':memory:')
cursor = conn.cursor()

# Create the Employees table
cursor.execute('''
CREATE TABLE Employees (
    EmployeeID INTEGER PRIMARY KEY,
    Name TEXT,
    Department TEXT
);
''')

# Insert sample data
cursor.execute("INSERT INTO Employees VALUES (1, 'Alice', 'Sales');")
cursor.execute("INSERT INTO Employees VALUES (2, 'Bob', 'Marketing');")
cursor.execute("INSERT INTO Employees VALUES (3, 'Alice', 'IT');")
cursor.execute("INSERT INTO Employees VALUES (4, 'David', 'Sales');")
cursor.execute("INSERT INTO Employees VALUES (5, 'Bob', 'Finance');")
conn.commit()

# SQL Query
sql_query = "SELECT Name FROM Employees GROUP BY Name HAVING COUNT(Name) > 1;"

# Execute query and fetch results
cursor.execute(sql_query)
results = cursor.fetchall()

# Display results using pandas
df = pd.DataFrame(results, columns=['Name'])
display(df)

# Explanation
print("\nExplanation: The `GROUP BY Name` clause groups all rows with the same name. The `HAVING COUNT(Name) > 1` filters these groups, returning only those names that appear more than once, effectively identifying duplicate names.")

# Close the connection
conn.close()
```

## Question 10: Second-Highest Salary

**Problem Statement:** Find the second-highest salary from the `Employees` table.

**Sample Table: Employees**

| EmployeeID | Name    | Salary |
|------------|---------|--------|
| 1          | Alice   | 60000  |
| 2          | Bob     | 50000  |
| 3          | Charlie | 75000  |
| 4          | David   | 62000  |
| 5          | Eve     | 70000  |

**Expected Output:**

| Salary |
|--------|
| 70000  |

```python
import sqlite3
import pandas as pd

# Connect to an in-memory SQLite database
conn = sqlite3.connect(':memory:')
cursor = conn.cursor()

# Create the Employees table
cursor.execute('''
CREATE TABLE Employees (
    EmployeeID INTEGER PRIMARY KEY,
    Name TEXT,
    Salary INTEGER
);
''')

# Insert sample data
cursor.execute("INSERT INTO Employees VALUES (1, 'Alice', 60000);")
cursor.execute("INSERT INTO Employees VALUES (2, 'Bob', 50000);")
cursor.execute("INSERT INTO Employees VALUES (3, 'Charlie', 75000);")
cursor.execute("INSERT INTO Employees VALUES (4, 'David', 62000);")
cursor.execute("INSERT INTO Employees VALUES (5, 'Eve', 70000);")
conn.commit()

# SQL Query
sql_query = "SELECT DISTINCT Salary FROM Employees ORDER BY Salary DESC LIMIT 1 OFFSET 1;"

# Execute query and fetch results
cursor.execute(sql_query)
results = cursor.fetchall()

# Display results using pandas
df = pd.DataFrame(results, columns=['Salary'])
display(df)

# Explanation
print("\nExplanation: This query first selects `DISTINCT Salary` values to handle cases with identical salaries. It then `ORDER BY Salary DESC` to sort salaries from highest to lowest. `LIMIT 1` fetches only one row, and `OFFSET 1` skips the first row (the highest salary), effectively giving us the second-highest salary.")

# Close the connection
conn.close()
```
