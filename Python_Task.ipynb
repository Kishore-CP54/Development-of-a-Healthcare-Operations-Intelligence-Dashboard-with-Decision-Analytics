## Question 1: Reverse a string

### Problem Statement
Write a Python function to reverse a given string.

### Example
- **Input:** `s = "hello"`
- **Output:** `olleh`

### Explanation
We can reverse a string by iterating through it from the end to the beginning and appending each character to a new string, or by using Python's slicing feature.

```python
# Function to reverse a string
def reverse_string(s):
    reversed_s = ""
    # Iterate through the string from the last character to the first
    for i in range(len(s) - 1, -1, -1):
        reversed_s += s[i] # Append each character to the new string
    return reversed_s

# Example usage
input_string = "hello"
output_string = reverse_string(input_string)
print(f"Original string: {input_string}")
print(f"Reversed string: {output_string}")

# Another way using slicing (more Pythonic but for beginners, loop is good to show understanding)
# def reverse_string_slicing(s):
#     return s[::-1]
# print(reverse_string_slicing("world"))
```

## Question 2: Check whether a number is prime

### Problem Statement
Write a Python function that takes an integer as input and returns `True` if the number is prime, and `False` otherwise.

### Example
- **Input:** `n = 7`
- **Output:** `True`
- **Input:** `n = 10`
- **Output:** `False`

### Explanation
A prime number is a natural number greater than 1 that has no positive divisors other than 1 and itself. We can check for primality by iterating from 2 up to the square root of the number and checking for divisibility. If any number in this range divides `n` evenly, then `n` is not prime.

```python
# Function to check if a number is prime
def is_prime(n):
    # Prime numbers must be greater than 1
    if n <= 1:
        return False
    # Iterate from 2 up to n-1 (inclusive of 2)
    # If n is divisible by any number in this range, it's not prime
    for i in range(2, n):
        if n % i == 0:
            return False
    # If no divisors are found, the number is prime
    return True

# Example usage
number1 = 7
print(f"Is {number1} prime? {is_prime(number1)}")

number2 = 10
print(f"Is {number2} prime? {is_prime(number2)}")

number3 = 2
print(f"Is {number3} prime? {is_prime(number3)}")

number4 = 1
print(f"Is {number4} prime? {is_prime(number4)}")
```

## Question 3: Find the largest number in a list

### Problem Statement
Write a Python function to find the largest number in a given list of numbers.

### Example
- **Input:** `numbers = [3, 1, 9, 4, 2]`
- **Output:** `9`

### Explanation
We can find the largest number by initializing a variable with the first element of the list and then iterating through the rest of the list, updating the variable whenever a larger number is found.

```python
# Function to find the largest number in a list
def find_largest_number(numbers):
    # Handle empty list case
    if not numbers:
        return None # Or raise an error, depending on requirements

    # Assume the first element is the largest initially
    largest = numbers[0]
    # Iterate through the rest of the list
    for number in numbers[1:]:
        # If the current number is greater than the current largest, update largest
        if number > largest:
            largest = number
    return largest

# Example usage
my_list1 = [3, 1, 9, 4, 2]
print(f"The largest number in {my_list1} is: {find_largest_number(my_list1)}")

my_list2 = [10, 5, 20, 15]
print(f"The largest number in {my_list2} is: {find_largest_number(my_list2)}")

my_list3 = []
print(f"The largest number in {my_list3} is: {find_largest_number(my_list3)}")
```

## Question 5: Count vowels in a string

### Problem Statement
Write a Python function to count the number of vowels (a, e, i, o, u) in a given string. The function should be case-insensitive.

### Example
- **Input:** `s = "Hello World"`
- **Output:** `3` (e, o, o)

### Explanation
We can iterate through each character of the string, convert it to lowercase, and check if it is present in a predefined set of vowels. If it is, we increment a counter.

```python
# Function to count vowels in a string
def count_vowels(s):
    vowels = "aeiou"
    vowel_count = 0
    # Convert the string to lowercase to handle case-insensitivity
    s_lower = s.lower()

    # Iterate through each character in the string
    for char in s_lower:
        # Check if the character is a vowel
        if char in vowels:
            vowel_count += 1 # Increment the counter if it's a vowel
    return vowel_count

# Example usage
text1 = "Hello World"
print(f"Number of vowels in '{text1}': {count_vowels(text1)}")

text2 = "Python Programming"
print(f"Number of vowels in '{text2}': {count_vowels(text2)}")

text3 = "AEIOU"
print(f"Number of vowels in '{text3}': {count_vowels(text3)}")

text4 = "Rhythm"
print(f"Number of vowels in '{text4}': {count_vowels(text4)}")
```

## Question 6: Find factorial of a number

### Problem Statement
Write a Python function to calculate the factorial of a given non-negative integer.

### Example
- **Input:** `n = 5`
- **Output:** `120` (since 5! = 5 * 4 * 3 * 2 * 1 = 120)

### Explanation
The factorial of a non-negative integer `n`, denoted by `n!`, is the product of all positive integers less than or equal to `n`. The factorial of 0 is 1. We can calculate this using a loop.

```python
# Function to calculate the factorial of a number
def factorial(n):
    # Base case: factorial of 0 is 1
    if n == 0:
        return 1
    # Handle negative numbers (factorials are not defined for negative integers)
    elif n < 0:
        return "Factorial is not defined for negative numbers"
    # Calculate factorial for positive numbers
    else:
        result = 1
        # Multiply result by each integer from 1 to n
        for i in range(1, n + 1):
            result *= i
        return result

# Example usage
num1 = 5
print(f"The factorial of {num1} is: {factorial(num1)}")

num2 = 0
print(f"The factorial of {num2} is: {factorial(num2)}")

num3 = 7
print(f"The factorial of {num3} is: {factorial(num3)}")

num4 = -3
print(f"The factorial of {num4} is: {factorial(num4)}")
```

## Question 7: Fibonacci series

### Problem Statement
Write a Python function to generate the Fibonacci series up to `n` terms. The Fibonacci sequence is a series where the next term is the sum of the previous two terms. The first two terms are 0 and 1.

### Example
- **Input:** `n = 7`
- **Output:** `[0, 1, 1, 2, 3, 5, 8]`

### Explanation
We can start with the first two terms (0 and 1) and then use a loop to generate subsequent terms by adding the previous two. We store these terms in a list.

```python
# Function to generate Fibonacci series up to n terms
def fibonacci_series(n):
    fib_list = []
    # Handle edge cases for n=0 and n=1
    if n <= 0:
        return fib_list # Empty list for non-positive n
    elif n == 1:
        fib_list.append(0)
        return fib_list
    else:
        # Initialize the first two terms
        a, b = 0, 1
        fib_list.append(a)
        fib_list.append(b)
        # Generate subsequent terms using a loop
        for _ in range(2, n):
            next_fib = a + b
            fib_list.append(next_fib)
            a = b
            b = next_fib
        return fib_list

# Example usage
terms1 = 7
print(f"Fibonacci series up to {terms1} terms: {fibonacci_series(terms1)}")

terms2 = 1
print(f"Fibonacci series up to {terms2} terms: {fibonacci_series(terms2)}")

terms3 = 0
print(f"Fibonacci series up to {terms3} terms: {fibonacci_series(terms3)}")

terms4 = 10
print(f"Fibonacci series up to {terms4} terms: {fibonacci_series(terms4)}")
```

## Question 8: Remove duplicates from a list

### Problem Statement
Write a Python function to remove duplicate elements from a list and return a new list with only unique elements, maintaining the original order as much as possible.

### Example
- **Input:** `my_list = [1, 2, 2, 3, 4, 3, 5]`
- **Output:** `[1, 2, 3, 4, 5]`

### Explanation
We can iterate through the original list and add each element to a new list only if it's not already present in the new list. Alternatively, a set can be used for faster checking of uniqueness, but to preserve order and return a list, the iteration method is straightforward for freshers.

```python
# Function to remove duplicates from a list
def remove_duplicates(input_list):
    unique_list = []
    # Iterate through each element in the original list
    for item in input_list:
        # If the item is not already in the unique_list, add it
        if item not in unique_list:
            unique_list.append(item)
    return unique_list

# Example usage
list1 = [1, 2, 2, 3, 4, 3, 5]
print(f"Original list: {list1}")
print(f"List after removing duplicates: {remove_duplicates(list1)}")

list2 = ["apple", "banana", "apple", "orange", "banana"]
print(f"Original list: {list2}")
print(f"List after removing duplicates: {remove_duplicates(list2)}")

list3 = []
print(f"Original list: {list3}")
print(f"List after removing duplicates: {remove_duplicates(list3)}")
```

## Question 9: Find the second-largest number

### Problem Statement
Write a Python function to find the second-largest number in a given list of numbers. Assume the list will always contain at least two unique numbers.

### Example
- **Input:** `numbers = [5, 2, 8, 1, 9]`
- **Output:** `8`

### Explanation
One way to find the second largest number is to first remove duplicates from the list, then sort the unique list in descending order, and pick the second element. Or, iterate through the list keeping track of the largest and second largest numbers.

```python
# Function to find the second largest number in a list
def find_second_largest(numbers):
    # First, remove duplicates to simplify the problem
    # Convert to set and back to list to get unique elements (order might change)
    # If order is crucial and elements should be removed maintaining original list order
    # then use the remove_duplicates function from Question 8.
    unique_numbers = []
    for num in numbers:
        if num not in unique_numbers:
            unique_numbers.append(num)

    # Handle cases where there are less than 2 unique numbers
    if len(unique_numbers) < 2:
        return "List does not have a second largest number (less than 2 unique elements)"

    # Sort the unique list in descending order
    unique_numbers.sort(reverse=True)

    # The second element in the sorted list will be the second largest
    return unique_numbers[1]

# Example usage
list1 = [5, 2, 8, 1, 9]
print(f"The second largest number in {list1} is: {find_second_largest(list1)}")

list2 = [10, 20, 30, 20]
print(f"The second largest number in {list2} is: {find_second_largest(list2)}")

list3 = [7, 7, 7]
print(f"The second largest number in {list3} is: {find_second_largest(list3)}")

list4 = [10, 5]
print(f"The second largest number in {list4} is: {find_second_largest(list4)}")
```

## Question 10: Count frequency of elements

### Problem Statement
Write a Python function to count the frequency of each element in a given list and return a dictionary where keys are the elements and values are their counts.

### Example
- **Input:** `my_list = [1, 2, 2, 3, 1, 4, 2]`
- **Output:** `{1: 2, 2: 3, 3: 1, 4: 1}`

### Explanation
We can iterate through the list. For each element, if it's already in our frequency dictionary, we increment its count. If not, we add it to the dictionary with a count of 1.

```python
# Function to count the frequency of elements in a list
def count_element_frequency(input_list):
    frequency_map = {}
    # Iterate through each element in the list
    for item in input_list:
        # If the element is already a key in the dictionary, increment its value
        if item in frequency_map:
            frequency_map[item] += 1
        # Otherwise, add the element to the dictionary with a value of 1
        else:
            frequency_map[item] = 1
    return frequency_map

# Example usage
list1 = [1, 2, 2, 3, 1, 4, 2]
print(f"Element frequencies in {list1}: {count_element_frequency(list1)}")

list2 = ["red", "blue", "green", "red", "blue", "red"]
print(f"Element frequencies in {list2}: {count_element_frequency(list2)}")

list3 = []
print(f"Element frequencies in {list3}: {count_element_frequency(list3)}")
```
